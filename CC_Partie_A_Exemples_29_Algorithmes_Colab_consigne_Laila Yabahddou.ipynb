# **Q1: **

.1 Présentation
La régression linéaire (Linear Regression) est un algorithme de la famille de l'apprentissage supervisé, plus précisément de la régression. Son objectif est de prédire une valeur numérique continue à partir d'une ou plusieurs variables explicatives.

Son principe consiste à rechercher la droite (ou l'hyperplan) qui représente le mieux la relation entre les variables d'entrée et la variable cible.

Les données d'entrée sont constituées de variables explicatives (X) et de valeurs connues (y). Pendant l'apprentissage, l'algorithme estime les coefficients qui minimisent les erreurs entre les valeurs réelles et les valeurs prédites.

La prédiction obtenue est calculée selon la formule :

OÛ:
Y: Valeur prédite
beta_0: constante (interception)
beta_j: coefficient appris pour la variable (x_j)
(x_j) : valeur de la variable explicative
(p): nombre de variables explicatives




Les coefficients (beta_0,beta_1,beta_p) sont des paramètres appris automatiquement. La régression linéaire classique ne nécessite pas d'hyperparamètre de régularisation.

Avantages :

Algorithme simple, rapide et facile à interpréter.

Fonctionne bien lorsque les relations entre les variables sont approximativement linéaires.

Limites :

Ne représente pas correctement les relations fortement non linéaires.

Peut être sensible aux valeurs aberrantes et à la multicolinéarité.

1.2 Schéma explicatif

Le schéma suivant présente les principales étapes de la régression linéaire.



````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Prédictions
predictions = modele.predict(Xr_test)

# 1. Valeurs réelles vs valeurs prédites
plt.figure(figsize=(8, 5))
plt.scatter(yr_test, predictions)

plt.plot(
    [min(yr_test), max(yr_test)],
    [min(yr_test), max(yr_test)],
    'r--'
)

plt.xlabel("Valeurs réelles")
plt.ylabel("Valeurs prédites")
plt.title("Régression linéaire : Réel vs Prédit")
plt.grid()
plt.show()


# 2. Graphique des résidus
residus = yr_test - predictions

plt.figure(figsize=(8, 5))
plt.scatter(predictions, residus)
plt.axhline(y=0, linestyle='--')

plt.xlabel("Valeurs prédites")
plt.ylabel("Résidus")
plt.title("Régression linéaire : Analyse des résidus")
plt.grid()
plt.show()


# 3. Distribution des erreurs
plt.figure(figsize=(8, 5))
sns.histplot(residus, kde=True)

plt.xlabel("Erreurs de prédiction")
plt.ylabel("Fréquence")
plt.title("Distribution des erreurs - Régression linéaire")
plt.show()
````

### 2.1 Présentation

La régression Ridge (*Ridge Regression*) est un algorithme d'apprentissage supervisé appartenant à la famille de la régression. Elle constitue une extension de la régression linéaire classique.

Son objectif est de prédire une valeur numérique continue tout en limitant les coefficients du modèle afin de réduire le surapprentissage (*overfitting*).

Son principe consiste à minimiser les erreurs de prédiction tout en ajoutant une pénalité sur la taille des coefficients. Cette pénalité permet de limiter leur importance et de stabiliser le modèle.

Le critère mathématique utilisé est :

\[
J(\beta)=\sum_{i=1}^{n}(y_i-\hat{y}_i)^2+
\alpha\sum_{j=1}^{p}\beta_j^2
\]

Où :
- \(J(\beta)\) : fonction objectif à minimiser ;
- \(n\) : nombre d'observations ;
- \(y_i\) : valeur réelle ;
- \(\hat{y}_i\) : valeur prédite ;
- \(\beta_j\) : coefficient associé à la variable \(j\) ;
- \(p\) : nombre de variables explicatives ;
- \(\alpha\) : hyperparamètre de régularisation Ridge.

Les coefficients \(\beta_j\) sont des paramètres appris automatiquement. En revanche, \(\alpha\) est un hyperparamètre choisi par l'utilisateur. Dans notre code, sa valeur est fixée à 1.0.

**Avantages :**
- Réduit le surapprentissage grâce à la régularisation.
- Gère mieux les problèmes de multicolinéarité entre variables explicatives.

**Limites :**
- Nécessite de choisir correctement la valeur de l'hyperparamètre alpha.
- Réduit les coefficients sans les annuler complètement, ce qui ne réalise pas une sélection automatique des variables.

### 2.4 Analyse de l'expérience Python

Le code utilise `Ridge` de Scikit-learn associé à `StandardScaler()` dans un pipeline.

- **Entrées :** `Xr_ent` représente les variables explicatives d'entraînement et `yr_ent` représente les valeurs cibles.
- **Standardisation :** `StandardScaler()` transforme les variables explicatives pour les centrer et les réduire, afin de leur donner une échelle comparable.
- **Apprentissage :** `Ridge(alpha=1.0)` entraîne le modèle en utilisant une régularisation L2.
- **Hyperparamètre :** `alpha=1.0` contrôle l'intensité de la régularisation. Une valeur plus élevée augmente la pénalisation des coefficients.
- **Prédiction :** le pipeline applique automatiquement la standardisation aux données de test avant de calculer les prédictions.
- **Sortie :** la MAE mesure l'erreur absolue moyenne entre les valeurs réelles et les prédictions.

**Interprétation :** cette expérience permet d'évaluer les performances du modèle Ridge sur les données de test. La valeur de la MAE indique l'erreur moyenne de prédiction dans l'unité de la variable cible. Toutefois, sans comparer cette valeur à celle de la régression linéaire classique, on ne peut pas affirmer que Ridge est meilleur. Il est également nécessaire de tester plusieurs valeurs de alpha pour déterminer la régularisation la plus adaptée.

````python
## schema
import matplotlib.pyplot as plt
import numpy as np

# Coefficients fictifs pour illustrer l'effet de Ridge
noms = ["Variable 1", "Variable 2", "Variable 3", "Variable 4"]

coefficients_lineaire = [8, -6, 5, 3]
coefficients_ridge = [5.5, -4.2, 3.6, 2.1]

positions = np.arange(len(noms))
largeur = 0.35

plt.figure(figsize=(9, 5))

plt.bar(positions - largeur/2,
        coefficients_lineaire,
        largeur,
        label="Régression linéaire")

plt.bar(positions + largeur/2,
        coefficients_ridge,
        largeur,
        label="Régression Ridge")

plt.axhline(0, linewidth=0.8)
plt.xticks(positions, noms)
plt.ylabel("Valeur des coefficients")
plt.title("Effet de la régularisation Ridge")
plt.legend()
plt.grid(axis="y")
plt.show()
````

````python
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error

modele = make_pipeline(
    StandardScaler(),
    Ridge(alpha=1.0)
).fit(Xr_ent, yr_ent)

predictions = modele.predict(Xr_test)

print('MAE:', mean_absolute_error(yr_test, predictions))
````

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Prédictions Ridge
predictions_ridge = modele.predict(Xr_test)

# 1. Valeurs réelles vs valeurs prédites
plt.figure(figsize=(8, 5))

plt.scatter(yr_test, predictions_ridge)

plt.plot(
    [min(yr_test), max(yr_test)],
    [min(yr_test), max(yr_test)],
    'r--'
)

plt.xlabel("Valeurs réelles")
plt.ylabel("Valeurs prédites")
plt.title("Ridge : Réel vs Prédit")
plt.grid()
plt.show()


# 2. Graphique des résidus
residus_ridge = yr_test - predictions_ridge

plt.figure(figsize=(8, 5))
plt.scatter(predictions_ridge, residus_ridge)
plt.axhline(y=0, linestyle='--')

plt.xlabel("Valeurs prédites")
plt.ylabel("Résidus")
plt.title("Ridge : Analyse des résidus")
plt.grid()
plt.show()


# 3. Distribution des erreurs
plt.figure(figsize=(8, 5))
sns.histplot(residus_ridge, kde=True)

plt.xlabel("Erreurs de prédiction")
plt.ylabel("Fréquence")
plt.title("Distribution des erreurs - Ridge")
plt.show()
````

.1 Présentation
La régression Lasso (Least Absolute Shrinkage and Selection Operator) appartient à la famille de la régression supervisée. Elle permet de prédire une valeur numérique continue et de sélectionner automatiquement les variables les plus importantes.

Son principe consiste à minimiser les erreurs de prédiction en ajoutant une pénalité sur la valeur absolue des coefficients.

Formule : \[
J_beta=sum_{i=1}^{n}(y_i- y_i)^2+
alpha\sum_{j=1}^{p}|beta_j|]

 avec: Yi: valeur réelle
       Ÿ : valeur predite
       Beta_j: coefficient appris
       alpha: hyperparamètre de régularisation

Avantages : sélection automatique des variables ; réduction du surapprentissage.

Limites : sensible au choix de alpha ; peut supprimer des variables utiles lorsque plusieurs variables sont corrélées.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
````

4.1 Présentation
La forêt aléatoire (Random Forest Regressor) appartient à la famille de la régression supervisée. Elle combine plusieurs arbres de décision pour prédire une valeur numérique.

Son principe consiste à construire plusieurs arbres sur des échantillons différents des données, puis à calculer la moyenne de leurs prédictions.

Avantages : réduit la variance par rapport à un arbre isolé ; représente les relations complexes et non linéaires.

Limites : nécessite davantage de ressources de calcul ; moins facile à interpréter qu'un arbre unique.

4.2 Schéma explicatif

````python
import matplotlib.pyplot as plt

arbres = ["Arbre 1", "Arbre 2", "Arbre 3"]
predictions = [100, 120, 110]

plt.figure(figsize=(7,4))
plt.bar(arbres, predictions)
plt.axhline(sum(predictions)/len(predictions),
            linestyle="--", label="Moyenne finale")

plt.ylabel("Valeurs prédites")
plt.title("Principe de la forêt aléatoire")
plt.legend()
plt.show()
````

Commentaire : chaque arbre produit une prédiction indépendante. La forêt aléatoire calcule ensuite la moyenne des prédictions pour obtenir le résultat final.

Source : schéma illustratif original avec valeurs fictives.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
````

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
````
